## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset includes content metrics, search traffic, and the decline label, which makes it suited for this lane. The code below reports the dataset size, how common declining pages are, and how many rows lack keyword context.

```python
import pandas as pd
from pathlib import Path

cwd = Path.cwd()
possible_paths = [
    cwd / "data" / "raw" / "content_refresh_anonymized.csv",
    cwd / ".." / "data" / "raw" / "content_refresh_anonymized.csv",
    cwd / ".." / ".." / "data" / "raw" / "content_refresh_anonymized.csv",
    cwd / ".." / ".." / ".." / "data" / "raw" / "content_refresh_anonymized.csv",
]

file_path = None
for candidate in possible_paths:
    if candidate.exists():
        file_path = candidate.resolve()
        break

if file_path is None:
    raise FileNotFoundError(
        f"Could not find data/raw/content_refresh_anonymized.csv from cwd={cwd}\n"
        f"Tried paths: {[str(p) for p in possible_paths]}"
    )

print(f"Loading CSV from: {file_path}")

df = pd.read_csv(file_path)

total_rows = len(df)
ndeclining = (df["trend_direction"].str.lower() == "down").sum()
client_count = df["client_id"].nunique()
missing_keyword = df["search_volume"].isna().sum()

print(f"Total rows: {total_rows}")
print(f"Declining pages: {ndeclining} ({ndeclining / total_rows:.1%})")
print(f"Distinct clients: {client_count}")
print(f"Rows missing keyword search volume: {missing_keyword}")
```
